#### Ejemplo de if

```python
# 1. Estructura if simple con Operadores Relacionales
# Se utiliza para ejecutar un bloque de código solo si la condición es verdadera

edad = 20
if edad >= 18:
    print("Es mayor de edad.")
# Descripción de las líneas:
# Línea 1: Se asigna el valor entero 20 a la variable edad

# Línea 2: El condicional if evalúa si edad es mayor o igual (>=) a 18 (operador relacional)

# Línea 3: Si la condición es cierta, se imprime el mensaje gracias a la indentación

```

#### Ejemplo if else

```python
# 2. Estructura if-else con Operadores Aritméticos
# Permite elegir entre dos alternativas: una si la condición es verdadera y otra si es falsa

numero = 7
if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
# Descripción de las líneas:
# Línea 1: Se inicializa la variable numero con el valor 7

# Línea 2: Se usa el operador aritmético de módulo (%) para obtener el resto de la división entre 2 y compararlo con 0 usando el operador relacional de igualdad (==)

# Línea 3: Se ejecuta este bloque si el resto es cero (es decir, el número es par).
# Línea 4: La cláusula else define el camino alternativo si la condición del if falla

# Línea 5: Se imprime que es impar al no cumplirse la igualdad.
```

```python
# 2. Estructura if-else con Operadores Aritméticos
# Permite elegir entre dos alternativas: una si la condición es verdadera y otra si es falsa

numero = 8
if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
# Descripción de las líneas:
# Línea 1: Se inicializa la variable numero con el valor 7

# Línea 2: Se usa el operador aritmético de módulo (%) para obtener el resto de la división entre 2 y compararlo con 0 usando el operador relacional de igualdad (==)

# Línea 3: Se ejecuta este bloque si el resto es cero (es decir, el número es par).
# Línea 4: La cláusula else define el camino alternativo si la condición del if falla

# Línea 5: Se imprime que es impar al no cumplirse la igualdad.
```

#### Ejemplo if, elif, else

```python
# 3. Estructura if-elif-else con Operadores Lógicos
#Se utiliza cuando existen múltiples condiciones encadenadas. Los operadores lógicos (and, or, not) permiten evaluar varias comparaciones a la vez

año = 2024
numero = int(input("Ingrese un número: "))
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print("Es un año bisiesto.")
elif año % 100 == 0:
    print("No es un año bisiesto.")
else:
    print("No es un año bisiesto.")
# Descripción de las líneas:
# Línea 1: Se define la variable año

# Línea 2: La condición utiliza operadores aritméticos (%), relacionales (==, !=) y lógicos (and, or) para verificar las reglas de un año bisiesto

# Línea 3: Si toda la expresión lógica resulta verdadera (True), imprime que es bisiesto

# Línea 4 y 5: Si no se cumple ninguna de las condiciones lógicas, se ejecuta el bloque else
```

```python
# 3. Estructura if-elif-else con Operadores Lógicos
#Se utiliza cuando existen múltiples condiciones encadenadas. Los operadores lógicos (and, or, not) permiten evaluar varias comparaciones a la vez

año = 2024
numero = int(input("Ingrese un número: "))
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print("Es un año bisiesto.")
elif año % 100 == 0:
    print("No es un año bisiesto.")
else:
    print("No es un año bisiesto.")
# Descripción de las líneas:
# Línea 1: Se define la variable año

# Línea 2: La condición utiliza operadores aritméticos (%), relacionales (==, !=) y lógicos (and, or) para verificar las reglas de un año bisiesto

# Línea 3: Si toda la expresión lógica resulta verdadera (True), imprime que es bisiesto

# Línea 4 y 5: Si no se cumple ninguna de las condiciones lógicas, se ejecuta el bloque else
```

```python
# 3. Estructura if-elif-else con Operadores Lógicos
#Se utiliza cuando existen múltiples condiciones encadenadas. Los operadores lógicos (and, or, not) permiten evaluar varias comparaciones a la vez

año = 2024
numero = int(input("Ingrese un número: "))
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print("Es un año bisiesto.")
elif año % 100 == 0:
    print("No es un año bisiesto.")
else:
    print("No es un año bisiesto.")
# Descripción de las líneas:
# Línea 1: Se define la variable año

# Línea 2: La condición utiliza operadores aritméticos (%), relacionales (==, !=) y lógicos (and, or) para verificar las reglas de un año bisiesto

# Línea 3: Si toda la expresión lógica resulta verdadera (True), imprime que es bisiesto

# Línea 4 y 5: Si no se cumple ninguna de las condiciones lógicas, se ejecuta el bloque else
```

```python
# 4. Ejemplo Completo: Cálculo de Descuentos (Aritméticos y Relacionales)
#Este ejemplo muestra cómo determinar un valor basado en rangos específicos utilizando elif

cantidad = 12
if cantidad >= 10:
    descuento = 0.20
elif cantidad >= 5:
    descuento = 0.10
else:
    descuento = 0

total = 1000 * (1 - descuento)
print(f"Total a pagar: {total}")
#Descripción de las líneas:
#Línea 1: Se establece la cantidad comprada

#Línea 2: Se evalúa si la cantidad es mayor o igual a 10 (primer nivel de descuento)

#Línea 4: elif (abreviatura de else if) introduce una segunda condición si la primera fue falsa, verificando si la cantidad es al menos 5

#Línea 6: else captura cualquier caso menor a 5 artículos, asignando 0 descuento

#Línea 8: Realiza un cálculo aritmético multiplicando el precio base por el factor de descuento

#Línea 9: Imprime el resultado final utilizando un formato de cadena
```
